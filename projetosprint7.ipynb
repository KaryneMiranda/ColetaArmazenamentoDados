```python
print(result_7.shape, '\n')
print(result_7.columns.tolist(), '\n')
print(result_7.dtypes, '\n')
print(result_7.info(), '\n')
print(result_7.sample(10), '\n')
print(result_7.isna().sum(), '\n')
```

```python
# Convertendo o campo start_ts para datetime
result_7['start_ts'] = pd.to_datetime(result_7['start_ts'])
print(result_7.info())
```

```python
# Criando uma nova coluna para dia da semana
result_7['day_of_week'] = result_7['start_ts'].dt.day_name()
print(result_7['day_of_week'].unique())
```

```python
# Verificar registros quando o clima está ruim
bad_weather = result_7[result_7['weather_conditions'] == 'Bad']
print(bad_weather.count(), '\n')

# Verificar registros quando o clima está bom
good_weather = result_7[result_7['weather_conditions'] == 'Good']
print(good_weather.count(), '\n')
```

```python
# Verificar a duração média das corridas em clima ruim
mean_bad_weather = bad_weather['duration_seconds'].mean()
print('média das corridas em clima ruim')
print(mean_bad_weather, '\n')

# Verificar a duração média das corridas em clima bom
mean_good_weather = good_weather['duration_seconds'].mean()
print('média das corridas em clima bom')
print(mean_good_weather, '\n')
```

```python
# Definindo as hipóteses
print("=== TESTE DE HIPÓTESES ===")
print("Hipótese a ser testada: 'A duração média dos passeios do Loop para O'Hare muda nos sábados chuvosos'")
print()
print("H0 (Hipótese Nula): A duração média é IGUAL em dias de clima ruim e bom")
print("H0: μ_ruim = μ_bom")
print()
print("H1 (Hipótese Alternativa): A duração média é DIFERENTE em dias de clima ruim e bom") 
print("H1: μ_ruim ≠ μ_bom")
print()
print("Nível de significância (α): 0.05")
```

```python
alpha = 0.05

# Use os arrays completos, não as médias!
duracao_clima_ruim = bad_weather['duration_seconds']
duracao_clima_bom = good_weather['duration_seconds']

# Agora o teste t funcionará corretamente
results = stats.ttest_ind(duracao_clima_ruim, duracao_clima_bom, equal_var=False)

print(f'Estatística t: {results.statistic}')
print(f'Valor-p: {results.pvalue}')

if results.pvalue < alpha:
    print("Rejeitamos a hipótese nula - há diferença significativa na duração")
else:
    print("Não podemos rejeitar a hipótese nula")
```

```python
# As hipóteses foram formadas com base na questão, então foi definido que a hipótese nula seria aquela em que mesmo em dias ruins a duração média não muda em relação aos dias bons.
# A hipótese alternativa é aquela que resulta na diferença da hipótese nula.

# Critérios em relação a lógica das hipóteses acima:
#Teste bilateral porque você quer detectar qualquer diferença (não apenas se uma é maior)
#Faz sentido prático: clima ruim pode aumentar OU diminuir tempo (trânsito vs. cancelamentos)

# Critérios em nível de significância (alpha: 0.05):
#Padrão em análise de dados de negócios
#Equilibra rigor científico com praticidade
#Aceita 5% de chance de erro Tipo I (rejeitar H0 quando é verdadeira)

# Critérios em testes estatísticos escolhidos (Teste t de Student para amostras independentes):
#Amostras independentes: clima ruim vs. bom são grupos separados
#Comparação de médias: queremos comparar duração média entre grupos
#Variâncias diferentes: equal_var=False (Teste de Welch)
```
